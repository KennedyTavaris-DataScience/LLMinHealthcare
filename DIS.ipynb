```python
import pandas as pd
import ollama


def process_excel_to_csv(file_path, output_file):
    df = pd.read_excel(file_path)
    
    results = []
    

    for index, row in df.iterrows():
        prompt = row[0] 
        
        if pd.notna(prompt): 
            healthcare_prompt = f"As a doctor and medical professional, answer the following question in a professional and helpful manner: {prompt}"
            
            response = ollama.generate(model='llama3.2', prompt=healthcare_prompt)
            
            results.append({"Prompt": prompt, "Response": response['response']})
    
    response_df = pd.DataFrame(results)
    response_df.to_csv(output_file, index=False)


file_path = '/Users/kennedytavaris/Desktop/Paraphrased query.xlsx'  # Input file
output_file = '/Users/kennedytavaris/Desktop/responsesF.csv'  # Output CSV file


process_excel_to_csv(file_path, output_file)

```
